```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from tqdm import tqdm

# Step 1: Define the neural network architecture without dropout
class MLPWithMomentum(nn.Module):
    def __init__(self, input_size, hidden_size1, hidden_size2, output_size):
        super(MLPWithMomentum, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size1)  # first linear layer
        self.relu1 = nn.ReLU()  # relu
        self.fc2 = nn.Linear(hidden_size1, hidden_size2)  # second linear layer
        self.relu2 = nn.ReLU()  # relu
        self.fc3 = nn.Linear(hidden_size2, output_size)  # last linear layer

    def forward(self, x):
        x = self.relu1(self.fc1(x))
        x = self.relu2(self.fc2(x))
        x = self.fc3(x)
        return x

# Step 2: Load the CIFAR-10 dataset
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

train_dataset = datasets.CIFAR10(root='./data', train=True, download=True, transform=transform)
test_dataset = datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)

# Step 3: Create DataLoader for training and testing
batch_size = 64

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

# Step 4: Instantiate the model, loss function, and optimizer
input_size = 32 * 32 * 3  # input size for CIFAR-10
hidden_size1 = 256
hidden_size2 = 128
output_size = 10


model = MLPWithMomentum(input_size, hidden_size1, hidden_size2, output_size)
criterion = nn.CrossEntropyLoss()

momentum = [0.1, 0.9]
# Step 5: Training the model
num_epochs = 10

for mt in momentum:
    optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=mt)  # Use SGD with momentum
    print("------------------- MOMENTUM IS {} ----------------------------".format(mt))

    for epoch in range(num_epochs):
        model.train()  # training mode here
        running_loss = 0.0
        total_train = 0
        correct_train = 0

        for inputs, labels in tqdm(train_loader, desc=f'Epoch {epoch + 1}/{num_epochs}'):
            inputs = inputs.view(inputs.size(0), -1)  # Flatten the input images
            optimizer.zero_grad()  # zero the gradient
            outputs = model(inputs)
            loss = criterion(outputs, labels)  # calculate the loss
            loss.backward()
            optimizer.step()

            running_loss += loss.item()

            _, predicted = torch.max(outputs.data, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()

        train_accuracy = correct_train / total_train
        print(f'Training Loss: {running_loss / len(train_loader)}, Training Accuracy: {train_accuracy * 100:.2f}%')

    # Step 6: Testing the model
    model.eval()  # Eval mode here
    correct = 0
    total = 0

    with torch.no_grad():
        for inputs, labels in tqdm(test_loader, desc='Testing'):
            inputs = inputs.view(inputs.size(0), -1)
            outputs = model(inputs)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    accuracy = correct / total
    print(f'Test Accuracy: {accuracy * 100:.2f}%')
```
